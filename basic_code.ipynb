```python
year = int(input("Enter year: "))
month = int(input("Enter month (1-12): "))

if month < 1 or month > 12:
    print("Invalid Month")
else:
    if month in [1,3,5,7,8,10,12]:
        print("31 Days")
    elif month in [4,6,9,11]:
        print("30 Days")
    else:
        if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
            print("29 Days")
        else:
            print("28 Days")
```

```python
units = int(input("Enter Units Consumed: "))

bill = 0

if units <= 100:
    bill = 0
elif units <= 300:
    bill = (units - 100) * 5
else:
    bill = 200 * 5 + (units - 300) * 10

if units > 500:
    surcharge = bill * 0.10
    bill += surcharge

print("Electricity Bill = ₹", bill)
```

```python
num = int(input("Enter Number: "))

product = 1

while num > 0:
    digit = num % 10
    product *= digit
    num = num // 10

print("Product =", product)
```

```python
num = int(input("Enter Number: "))

sum = 0

for i in range(1, num + 1):
    if num % i == 0:
        sum += i

print("Sum of Factors =", sum)
```

```python
num = int(input("Enter Decimal Number: "))

binary = ""

while num > 0:
    binary = str(num % 2) + binary
    num = num // 2

print("Binary =", binary)
```

```python
num = int(input("Enter Number: "))

largest = 0

while num > 0:
    digit = num % 10

    if digit > largest:
        largest = digit

    num = num // 10

print("Largest Digit =", largest)
```

```python
n = int(input("Enter N: "))

print("Prime Numbers:")

for num in range(2, n + 1):
    prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```
